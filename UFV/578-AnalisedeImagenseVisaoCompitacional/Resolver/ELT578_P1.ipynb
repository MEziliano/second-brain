**2) Importando bibliotecas**

```python
import numpy as np
import cv2
```

**3) Lendo e exibindo uma imagem colorida (BGR)**

```python
from google.colab.patches import cv2_imshow
img = cv2.imread('/content/drive/MyDrive/ELT578/SEMANA_1/flower.jpeg')
cv2_imshow(img)
```

**4) Explorando as propriedades da imagem**

```python
print ('Dimensoes da Imagem Flor (linhas, colunas e bandas) = ', img.shape)

print ('Tamanho (px) da Imagem Flor = ', img.size)

print ('Valores das bandas B(0)G(1)R(2) na linha 100 e coluna 100 da Flor = ', img[99,99,])

print ('Valor da banda vermelho na linha 100 e coluna 100 da Flor = ', img[99,99,1])
```

**5) Lendo a mesma imagem em tons de cinza**

```python
img2 = cv2.imread('/content/drive/MyDrive/ELT578/SEMANA_1/flower.jpeg', 0) #cv2.IMREAD_GRAYSCALE
cv2_imshow(img2)
```

**6) Propriedades da imagem em tons de cinza**

```python
print ('Dimensoes da Imagem Flor (linhas, colunas e bandas) = ', img2.shape)
print ('Tamanho (px) da Imagem Flor = ', img2.size)
print ('Valores das bandas B(0)G(1)R(2) na linha 100 e coluna 100 da Flor = ', img2[99,99,])
```

**7) Visualizando com Matplotlib (corrigindo BGR®RGB)**

```python
from matplotlib import pyplot as plt

#plt.imshow(img)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.xticks([]), plt.yticks([]) # esconder os valors de Ox e Oy #esconder o eixo
plt.show()

cv2.imwrite('/content/drive/MyDrive/ELT578/SEMANA_1/flowerGray.jpeg', cv2.cvtColor(img, cv2.COLOR_BGR2GRAY))
```

**8) Comparando RGB vs Gray lado a lado**

```python
#Visualizar as duas imagens ao mesmo tempo
titulos = ['Flor RGB', 'Flor Gray']
imagens = [img, img2]

#0,1 em uma lista
for i in range(2): # i varia de 0 a 1
    plt.subplot (1, 2, i+1), plt.imshow(cv2.cvtColor(imagens[i], cv2.COLOR_BGR2RGB))
    plt.title(titulos[i])
    plt.xticks([]), plt.yticks([])
plt.show()
```
