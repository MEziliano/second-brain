```python
import numpy as np
import cv2
from google.colab.patches import cv2_imshow
```

```python
from matplotlib import pyplot as plt
img = cv2.imread('/content/drive/MyDrive/ELT578/SEMANA_1/Fuji.bmp')


#Qual é o tamanho de armazenamento teórico da imagem?

# Inicialmente, vamos determinar a profundidade do pixel da imagem
print(img.dtype)
print(img.shape)
#Considerando que a imagem tem 2048 linhas e 946 colunas, temos:
npx=436*308

#Pela profundidade do pixel, temos que são necessários 8 bits para armazenar esse valor ou 1 byte,
#logo o número de bytes na imagem coincide com número de pixels, se a imagem possui apenas 1 banda (imagem em escala de cinza).
#Sabendo que 1kB = 1024 bytes, temos:


#Considerando que cada byte de armazenamento é equivalente a 8 bits e que a imagem "Flor" possuí 3 bandas, ou seja, 24bits.
#O tamanho de armazenamento digital em kilobytes (1024bytes) da imagem é dado por:
size=npx*3/(1024)
print('O tamanho teórico da imagem é %1.3f kB' %size)

import os
print('O tamanho real da imagem é de %1.3f kB' % (os.path.getsize('/content/drive/MyDrive/ELT578/SEMANA_1/Fuji.bmp')/1024))

#O Truecolor de 24 bits utiliza 8 bits para representar cada uma das cores RGB. 2 elevado a 8 = 256 níveis que combinados geram 16 777 216 cores (256 × 256 × 256).
```

```python
#Salvando a imagem em formato JPEG com 3 diferentes níveis de qualidade
cv2.imwrite('/content/drive/MyDrive/ELT578/SEMANA_1/fuji25.jpeg', img, [int(cv2.IMWRITE_JPEG_QUALITY), 25])
cv2.imwrite('/content/drive/MyDrive/ELT578/SEMANA_1/fuji50.jpeg', img, [int(cv2.IMWRITE_JPEG_QUALITY), 50])
cv2.imwrite('/content/drive/MyDrive/ELT578/SEMANA_1/fuji75.jpeg', img, [int(cv2.IMWRITE_JPEG_QUALITY), 75])
cv2.imwrite('/content/drive/MyDrive/ELT578/SEMANA_1/fuji100.jpeg', img, [int(cv2.IMWRITE_JPEG_QUALITY), 100])


#Visualizar as 4 imagens ao mesmo tempo
img25 = cv2.imread('/content/drive/MyDrive/ELT578/SEMANA_1/fuji25.jpeg')
img50 = cv2.imread('/content/drive/MyDrive/ELT578/SEMANA_1/fuji50.jpeg')
img75 = cv2.imread('/content/drive/MyDrive/ELT578/SEMANA_1/fuji75.jpeg')
img100 = cv2.imread('/content/drive/MyDrive/ELT578/SEMANA_1/fuji100.jpeg')

title = ['Fuji 25', 'Fuji 50', 'Fuji 75', 'Fuji 100']
imagens = [img25, img50, img75, img100]

for i in list(range(4)):
    plt.subplot (2,2,i+1), plt.imshow(cv2.cvtColor(imagens[i], cv2.COLOR_BGR2RGB))
    plt.title(title[i])
    plt.xticks([]), plt.yticks([])
plt.show()

print(img25.shape)
print(img50.shape)
print(img75.shape)
print(img100.shape)

#As imagens foram salvas em quatro padrões de qualidade diferentes (25%, 50%, 75% e 100%), podemos observar que todas as imagens possuem a mesma resolução da imagem, bem como a mesma resolução de brilho (24 bits, dado que uma imagem RGB possuí 3 bandas).
#A mudança quantitativa é então observada a nível de tamanho de armazenamento (kBytes).
#Os formatos de imagem são padrões de codificação, compressão e definição que determinam o tipo de arquivo gráfico, como JPEG, PNG, BMP, GIF e TIFF.
#Muitas vezes, o formato escolhido é uma questão de equilíbrio entre a qualidade e o tamanho da imagem, de acordo com a aplicação desejada.
#Quanto menor a qualidade, mais informação de alta frequência é descartada, o que resulta em maior compressão e maior perda de detalhes.
```
