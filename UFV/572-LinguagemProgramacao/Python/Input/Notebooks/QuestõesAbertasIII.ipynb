Qual será o conteúdo da lista `resultados' após a execução do loop, e qual será o valor da variável `resultado_final`? Explique como o código funciona.

```python
# Exercício 4
numeros = [1,2,3,4,5,6]
resultado = []

for i in range(len(numeros)):
  if i % 2 == 0:
    resultado.append(numeros[i] **2)
  else:
    resultado.append(numeros[i] *2)
print(resultado)
resultado_final = sum(resultado)
print(resultado_final)
```

**Explicação:** O resultado será de: `[1,4,9,8,25,12]` para a variável `resultado`. E `59` para a variável `resultado_final`.

<br>
Pode se observar que o código inicializa uma lista de números `numeros = [1,2,3,4,5,6]` e uma lista vazia `resultado = []` que será preenchida durante a execução.

Em seguida, o código entra em um loop `for` que itera pelos índices da lista `numeros`. Para cada índice `i`:
- Se o índice `i` for par (`i % 2 == 0`), o código calcula o quadrado do número na posição `numeros[i]` (ou seja, `numeros[i] ** 2`) e adiciona esse valor à lista `resultado`.
- Se o índice `i` for ímpar (`else`), o código calcula o dobro do número na posição `numeros[i]` (ou seja, `numeros[i] * 2`) e adiciona esse valor à lista `resultado`.

Após o loop, a lista `resultado` é impressa. Por exemplo:
- Para `i=0` (par), `numeros[0]` é 1, `1**2 = 1`. `resultado` = `[1]`
- Para `i=1` (ímpar), `numeros[1]` é 2, `2*2 = 4`. `resultado` = `[1, 4]`
- Para `i=2` (par), `numeros[2]` é 3, `3**2 = 9`. `resultado` = `[1, 4, 9]`
- Para `i=3` (ímpar), `numeros[3]` é 4, `4*2 = 8`. `resultado` = `[1, 4, 9, 8]`
- Para `i=4` (par), `numeros[4]` é 5, `5**2 = 25`. `resultado` = `[1, 4, 9, 8, 25]`
- Para `i=5` (ímpar), `numeros[5]` é 6, `6*2 = 12`. `resultado` = `[1, 4, 9, 8, 25, 12]`

Finalmente, a variável `resultado_final` recebe a soma de todos os elementos da lista `resultado` (`sum(resultado)`) e esse valor também é impresso. Neste caso, `1 + 4 + 9 + 8 + 25 + 12 = 59`.
