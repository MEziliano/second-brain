## 3. Análise Univariada

```python
print("\t MEDIDAS DE TENDÊNCIA CENTRAL \n", '='*50)
data = df.select_dtypes(include="number")
for column in data:
  print("A mínima  de " f"{column}" " é de: " +f'{round(data[column].min()):.2f}')
  print("A mediana de " f"{column}" " é de: " +f'{round(data[column].median()):.2f}')
  print("A média   de "f"{column}"  " é de: " +f'{round(data[column].mean()):.2f}')
  print("A maxima  de " f"{column}" " é de: " +f'{round(data[column].max()):.2f}')

  print("--"*25)
```

```python
print("\t MEDIDDAS DE TENDÊNCIA VARIABILIDADE \n", '='*50)
data = df.select_dtypes(include="number")
for column in data:
  print("O Desvio Padrão de " f"{column}" " é de: " +f'{round(data[column].std()):.2f}')
  print("A Variancia de " f"{column}" " é de: " +f'{round(data[column].var()):.2f}')
  print("O Amplitude de " f"{column}" " é de: " +f'{round(data[column].max() - data[column].min()):.2f}')
  print("O coeficiente de Variação de " f"{column}" " é de: "+ f'{round(data[column].std() / data[column].mean()*100):.2f}')

  print("--"*25)
```

```python
for column in data:
  plt.figure(figsize=(18,6))
  sns.histplot(x=column, data=df, element="step")
  plt.axvline(data[column].mean(), 0, 40, c='orange', label='Mean')
  plt.axvline(np.mean(data[column])+ np.std(data[column]), 0, 40,c='darkorange', label='mean + std', ls= "solid")
  plt.axvline(np.mean(data[column]) - np.std(data[column]), 0, 40,c='darkorange', label='mean - std', ls= "solid")
  plt.axvline(np.mean(data[column])+ (np.std(data[column])*2), 0, 40,c='gold', label='mean + *2std', ls= "--")
  plt.axvline(np.mean(data[column]) - (np.std(data[column])*2), 0, 40,c='gold', label='mean - 2*std', ls= "--")
  plt.axvline(np.mean(data[column])+ (np.std(data[column])*3), 0, 40,c='gold', label='mean + 3*std', ls= ":")
  plt.axvline(np.mean(data[column]) - (np.std(data[column])*3), 0, 40,c='gold', label='mean - 3*std', ls= ":")
  plt.legend()
  plt.title(column, fontsize=20)
  plt.xlabel(column)

  plt.show()
```

```python
for column in data:
  plt.figure(figsize=(18,6))
  sns.boxplot(x=column, data=df)
  plt.axvline(data[column].mean(), 0, 40, c='orange', label='Mean')
  plt.axvline(np.mean(data[column])+ np.std(data[column]), 0, 40,c='darkorange', label='mean + std', ls= "solid")
  plt.axvline(np.mean(data[column]) - np.std(data[column]), 0, 40,c='darkorange', label='mean - std', ls= "solid")
  plt.axvline(np.mean(data[column])+ (np.std(data[column])*2), 0, 40,c='gold', label='mean + *2std', ls= "--")
  plt.axvline(np.mean(data[column]) - (np.std(data[column])*2), 0, 40,c='gold', label='mean - 2*std', ls= "--")
  plt.axvline(np.mean(data[column])+ (np.std(data[column])*3), 0, 40,c='gold', label='mean + 3*std', ls= ":")
  plt.axvline(np.mean(data[column]) - (np.std(data[column])*3), 0, 40,c='gold', label='mean - 3*std', ls= ":")
  plt.legend()
  plt.title(column, fontsize=20)
  plt.xlabel(column)

  plt.show()
```

## 3. Análise de Outliers

```python
# Verificando os quartis de distribuição. 
Q1=df[["BuildingArea", "YearBuilt", "Landsize"]].quantile(0.25)
Q3=df[["BuildingArea", "YearBuilt", "Landsize"]].quantile(0.75)
IQR=Q3-Q1
print("\n")
print("The Q1 is :\n", Q1)
print("The Q3 is :\n", Q3)
print("The IQR is :\n", IQR)
lower_bound, upper_bound  = Q1-1.5*IQR, Q3+1.5*IQR

print("The Normal Range is: \n", lower_bound, "\n","-"*25,"\n", upper_bound)
```

```python
def treat_outliers():
  median = np.median(df)
  df["trestbps"] = df["trestbps"].mask(df["trestbps"] > 170, median)
  median = df.loc[df["trestbps"]<170, "trestbps"].median()
  df.loc[df.trestbps > 170, "trestbps"] = np.nan

  df["chol"] = df["chol"].mask(df["chol"] > 350, median)
  median = df.loc[df["chol"]<350, "chol"].median()
  df.loc[df.chol > 350, "chol"] = np.nan

  df["thalach"] = df["thalach"].mask(df["thalach"] < 80, median)
  median = df.loc[df["thalach"]<80, "thalach"].median()
  df.loc[df.thalach < 80, "thalach"] = np.nan

  df["oldpeak"] = df["oldpeak"].mask(df["oldpeak"] > 4.0, median)
  median = df.loc[df["oldpeak"]< 4.0, "oldpeak"].median()
  df.loc[df.oldpeak > 4.0, "oldpeak"] = np.nan
  return df
```
